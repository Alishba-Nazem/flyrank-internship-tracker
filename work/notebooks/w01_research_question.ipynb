## 2. The question: decision, action, cost of a wrong call

Decision:
Which pages should an editor review first?

Action:
The editor can review the highest-priority pages and decide whether to refresh, expand, protect, prune, or monitor them.

Cost of a wrong call:
The editor may waste time on a page that did not need attention, or miss a page that needed review.

## 3. Quick look at the data (2-3 real numbers)

30,000 pages / 32 clients
13,152 declining + demand pages = 43.8%
Top 500 pages = 32.8% of impressions

```python
from pathlib import Path
import pandas as pd

candidates = [
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("../../data/raw/content_refresh_anonymized.csv"),
]
csv_path = next(p for p in candidates if p.exists())

df = pd.read_csv(csv_path)

n_pages = len(df)
n_clients = df["client_id"].nunique()
pages_per_client = df.groupby("client_id").size()

declining_with_demand = (
    (df["trend_direction"] == "down")
    & (df["impressions_90d"] >= 100)
)
n_declining_demand = int(declining_with_demand.sum())
pct_declining_demand = 100 * n_declining_demand / n_pages
impr_share_declining = (
    100
    * df.loc[declining_with_demand, "impressions_90d"].sum()
    / df["impressions_90d"].sum()
)

top500 = df["impressions_90d"].nlargest(500)
impr_share_top500 = 100 * top500.sum() / df["impressions_90d"].sum()

print("1. Inventory size")
print("   pages:", n_pages)
print("   clients:", n_clients)
print("   median pages per client:", int(pages_per_client.median()))
print("   max pages on one client:", int(pages_per_client.max()))

print("2. Declining with demand (down AND impressions_90d >= 100)")
print("   pages:", n_declining_demand)
print("   share of pages (%):", round(pct_declining_demand, 1))
print("   share of all impressions (%):", round(impr_share_declining, 1))

print("3. Concentration of search attention")
print("   top 500 pages, share of all impressions (%):", round(impr_share_top500, 1))
```
